```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/tmdb_movies.csv")
```

```python
df.shape
```

```python
df_clean = df.drop_duplicates(subset="id", keep="first").copy()
```

```python
df_clean.shape
```

```python
df_clean["id"].duplicated().sum()
```

```python
df_clean[["overview", "genres"]].isnull().sum()
```

```python
df_clean[["overview", "genres"]].isnull().any(axis=1).sum()
```

```python
df_clean = df_clean.dropna(subset=["overview", "genres"]).copy()
```

```python
df_clean[["overview", "genres"]].isnull().sum()
```

```python
df_clean.shape
```

```python
df_clean.isnull().sum()
```

```python
df_clean["genres"] = df_clean["genres"].apply(
    lambda x: "|".join(sorted(x.split("|")))
)
```
